```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np

# Cores da Identidade Visual
COLORS = {'primary': '#1A237E', 'secondary': '#00E5FF', 'accent': '#FF4081', 'bg': '#F4F7FA'}

print('✅ Ambiente configurado.')
```

### 1. Gráfico de Cascata (Waterfall): Erosão do Faturamento
Este gráfico mostra o caminho do dinheiro desde o clique na venda até o que realmente entra no caixa.

```python
fig_water = go.Figure(go.Waterfall(
    name = 'Vazamento', orientation = 'v',
    measure = ['relative', 'relative', 'relative', 'relative', 'total'],
    x = ['Faturamento Bruto', 'Descontos', 'Cancelados/Fraude', 'Devoluções', 'Entrada Real'],
    textposition = 'outside',
    text = ['R$ 20.5M', '-R$ 1.6M', '-R$ 2.4M', '-R$ 2.5M', 'R$ 14.0M'],
    y = [20526133, -1636799, -2407110, -2500436, 0],
    connector = {'line':{'color':'#666'}},
    increasing = {'marker':{'color': COLORS['secondary']}},
    decreasing = {'marker':{'color': COLORS['accent']}},
    totals = {'marker':{'color': COLORS['primary']}}
))

fig_water.update_layout(title='Jornada de Receita (Leakage Analysis)', template='plotly_white')
fig_water.show()
```

### 2. Análise de Estoque e SAC
Comparativo entre o volume de chamados e a satisfação do cliente por motivo.

```python
motivos = ['Entrega', 'Defeito', 'Tamanho', 'Pagamento']
tickets = [10765, 6509, 5360, 4281]
csat = [2.98, 2.81, 2.99, 2.98]

fig_sac = make_subplots(specs=[[{'secondary_y': True}]])

fig_sac.add_trace(go.Bar(x=motivos, y=tickets, name='Volume Tickets', marker_color=COLORS['primary']), secondary_y=False)
fig_sac.add_trace(go.Scatter(x=motivos, y=csat, name='CSAT (0-5)', mode='lines+markers', marker_color=COLORS['accent']), secondary_y=True)

fig_sac.update_layout(title='Matriz de Ineficiência: Atendimento')
fig_sac.update_yaxes(title_text='Volume', secondary_y=False)
fig_sac.update_yaxes(title_text='Nota CSAT', range=[0, 5], secondary_y=True)
fig_sac.show()
```
